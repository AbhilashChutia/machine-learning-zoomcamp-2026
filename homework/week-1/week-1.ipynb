```python
df = pd.read_csv('car_fuel_efficiency_2026.csv')
```

```python
# record counts
record_counts = df.shape[0]
record_counts
```

```python
# fuel types
fuel_counts = df['fuel_type'].nunique()
fuel_counts
```

```python
# missing values
missing_cols = (df.isnull().sum() > 0).sum()
missing_cols
```

```python
# max fuel efficiency - asia
max_efficiency = df[df['origin'] == 'Asia']['fuel_efficiency_mpg'].max()
max_efficiency
```

```python
# median value of horsepower
median_hp = df['horsepower'].median()

# most frequent value
mode_hp = df['horsepower'].mode()[0]

# fill missing values with the mode
df['horsepower_filled'] = df['horsepower'].fillna(mode_hp)

# new median
new_median_hp = df['horsepower_filled'].median()

print(f"Original Median: {median_hp}")
print(f"New Median: {new_median_hp}")
```

```python
# all cars from asia
cars_asia = df[
    df['origin'] == 'Asia'
]

# select vehicle_weight, model_year - first 7 values
X = cars_asia[
    ['vehicle_weight', 'model_year']
].iloc[:7].values

# compute matrix multiplication - X transpose multiplied by X
XT = X.T
XTX = XT.dot(X)

# invert XTX
XTX_inv = np.linalg.inv(XTX)

# create array
y = np.array(
    [1100, 1300, 800, 900, 1000, 1100, 1200]
)

# multiply XTX_inv with XT, then multiply by y
w = XTX_inv.dot(XT).dot(y)

print(w.sum())
```
